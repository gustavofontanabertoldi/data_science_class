```python
import numpy as np

np.random.seed(42)

# 15 temperaturas entre 50°C e 100°C
temperaturas = np.round(np.random.uniform(50, 100, 15), 2)

# Criando DataFrame
df = pd.DataFrame({
    "Temperatura": temperaturas
})

# Classificação
df["Classificação"] = np.select(
    [
        df["Temperatura"] < 60,
        df["Temperatura"].between(60, 85),
        df["Temperatura"] > 85
    ],
    [
        "Baixa",
        "Normal",
        "Alta"
    ],
    default="Indefinida"
)

# Contagem das categorias
contagem = df["Classificação"].value_counts()

print(df)
print("\nQuantidade por categoria:")
print(contagem)
```

```python
rng = np.random.default_rng()
list_consumo = rng.integers(100, 300, size=10, endpoint=True)
pecas = rng.integers(100, 200, size=10, endpoint=True)

df = pd.DataFrame({
    "equipamentos": ["maquina_1", "maquina_2", "maquina_3", "maquina_4", "maquina_5", "maquina_6", "maquina_7", "maquina_8", "maquina_9", "maquina_10"],
    "consumo": list_consumo,
    "peças": pecas,
})

df["eficiencia"] = df["peças"] / df["consumo"]

df_ordenado = df.sort_values(by="eficiencia", ascending=False)
print(df.round(2))
```

```python
rng = np.random.default_rng()
dados = rng.normal(loc=150, scale=10, size=24)
datas = pd.date_range(start="2026-08-07 00:00" ,periods=24, freq="h")

serie = pd.Series(data=dados, index=datas)
media = serie.mean()
desvio_padrao = serie.std()
valor_max = serie.max()

# Exibir os resultados das métricas
print(f"Média: {media:.2f} kWh")
print(f"Desvio Padrão: {desvio_padrao:.2f} kWh")
print(f"Valor Máximo: {valor_max:.2f} kWh")

# Plotar o gráfico
serie.plot(title="Consumo de Energia do Forno (24h)", ylabel="kWh", xlabel="Horário")
```

```python
import numpy as np
import pandas as pd

np.random.seed(42)

equipamentos = [f"Equipamento {i}" for i in range(1, 21)]
consumo = np.random.uniform(100, 200, 20).round(2)
producao = np.random.randint(100, 1000, 20)

df = pd.DataFrame({
    "Equipamentos": equipamentos,
    "Consumo": consumo,
    "Produção": producao
})

consumo_norm = (
    (consumo - np.min(consumo)) /
    (np.max(consumo) - np.min(consumo))
)

producao_norm = (
    (producao - np.min(producao)) /
    (np.max(producao) - np.min(producao))
)

df["Consumo_normalizado"] = consumo_norm
df["Produção_normalizado"] = producao_norm

print(df)
```

```python
rng = np.random.default_rng()
list_temp = rng.integers(60, 101, size=10, endpoint=True)
status_list = ["Operando", "Manutenção", "Parado"]
escolhas = ["Risco Alto", "Risco Médio"]

df = pd.DataFrame({
    "equipamentos": [f"equipamento_{i}" for i in range(1, 11)],
    "status": rng.choice(status_list, size=10),
    "temperatura": list_temp
})

condicoes = [
    (df["temperatura"] > 90) & (df["status"] != "Parado"),
    ((df["temperatura"] >= 80) & (df["temperatura"] <= 90))
]

df["Risco"] = np.select(condicoes, escolhas, default="Risco Baixo")

print(df)
```
