## 🧮 ndarray – Estrutura Central do NumPy
O `ndarray` (N-dimensional array) é a estrutura fundamental do NumPy. Ele representa uma coleção de elementos (números, strings, booleanos) organizados em múltiplas dimensões, armazenados de forma contínua na memória e com tipo de dado fixo.

### ✅ Exemplo:
```python
import numpy as np
a = np.array([[1, 2, 3], [4, 5, 6]])
print(a)
print("Dimensões:", a.ndim)
print("Formato:", a.shape)
```

## ➕ Operações Vetorizadas
O NumPy permite operações matemáticas diretas sobre arrays, sem necessidade de loops explícitos. Isso é chamado de vetorização e melhora o desempenho do código.

### ✅ Exemplo:
```python
a = np.array([1, 2, 3])
b = np.array([10, 20, 30])
print("Soma:", a + b)
print("Multiplicação:", a * b)
```
## 🧠 Funções Estatísticas
NumPy fornece funções como `mean`, `std`, `sum`, `min`, `max` (média,desvio padrão, soma total, mínimo e máximo, respectivamente), fundamentais para a análise descritiva de dados. Essas funções permitem realizar análises rápidas sobre o comportamento de variáveis numéricas, identificar padrões, detectar anomalias e compreender a dispersão dos dados. Por exemplo, ao analisar leituras de sensores em uma linha de produção, é possível calcular rapidamente a média de temperatura ou o desvio padrão de vibração de um motor.

### ✅ Exemplo:
```python
dados = np.array([10, 20, 30, 40])
print("Média:", np.mean(dados))
print("Desvio padrão:", np.std(dados))
print("Máximo:", np.max(dados))
```
## 📐 Álgebra Linear
A biblioteca NumPy, através do módulo numpy.linalg, oferece uma variedade de operações matriciais essenciais como multiplicações de matrizes, cálculo de determinantes, extração de autovalores e autovetores, além da resolução de sistemas lineares. Essas ferramentas são fundamentais não apenas na matemática pura, mas também em aplicações práticas como simulações físicas, modelos econômicos e algoritmos de aprendizado de máquina. Por exemplo, ao resolver um sistema de equações lineares que representa o equilíbrio de cargas em uma rede elétrica ou otimizar funções quadráticas em regressão multivariada, essas operações se tornam cruciais.



### ✅ Exemplo:
```python
A = np.array([[3, 1], [1, 2]])
B = np.array([[1], [2]])
x = np.linalg.solve(A, B)
print("Solução do sistema:", x)
```

## 🔁 Transformações
O NumPy oferece um conjunto poderoso de operações de transformação de arrays, como reshape (reformulação da estrutura dimensional), flatten (achatamento de arrays multidimensionais em um vetor unidimensional), transpose (transposição de linhas por colunas), além da concatenação de arrays por diferentes eixos. Essas transformações são indispensáveis para adaptar dados às exigências de entrada dos algoritmos, reorganizar estruturas para visualização ou preparar saídas para exportação. Por exemplo, em um sistema de visão computacional, uma imagem colorida em 3D pode precisar ser convertida em um vetor 1D (flatten) para alimentar uma rede neural densa. Em outra situação, sensores distribuídos em diferentes linhas de produção podem gerar vetores que precisam ser concatenados verticalmente (empilhamento) para análise conjunta.



### ✅ Exemplo:
```python
a = np.array([[1, 2], [3, 4], [5, 6]])
print("Transposta:", a.T)
print("Flatten:", a.flatten())
print("Reshape para (2, 3):", a.reshape(2, 3))
```

## 🔢 Geração de Dados
O NumPy disponibiliza diversas funções para geração eficiente de dados e estruturas numéricas. Entre elas estão: arange (criação de sequências com passo definido), linspace (distribuições lineares entre valores), zeros e ones (arrays preenchidos com 0 ou 1), identity (matriz identidade), e o módulo random (geração de números aleatórios com diferentes distribuições). Essas ferramentas permitem a construção rápida de vetores e matrizes utilizadas para simulações, testes e inicialização de algoritmos. Por exemplo, em um ambiente industrial, é possível gerar uma série temporal simulada de temperaturas com linspace, ou criar vetores de referência zerados para comparações com sensores usando zeros.

### ✅ Exemplo:
```python
print(np.linspace(0, 1, 5))
print(np.zeros((2, 2)))
print(np.random.rand(3))
```
## 🧊 Manipulação de Tipos de Dados
No NumPy, todos os elementos de um ndarray possuem o mesmo tipo de dado (por exemplo: int32, float64, bool, etc.), o que garante eficiência na alocação de memória e velocidade nos cálculos vetorizados. Ao criar um array, podemos especificar o tipo explicitamente usando o parâmetro dtype, ou ainda converter um array existente para outro tipo com o método .astype(). Isso é particularmente útil em contextos onde o controle de precisão e consumo de memória são críticos. Por exemplo, ao lidar com grandes conjuntos de imagens ou sinais industriais, pode-se preferir float32 em vez de float64 para reduzir o uso de memória.

### ✅ Exemplo:
```python
valores = np.array([1.2, 2.5, 3.1], dtype=np.float32)
print(valores)
print("Tipo de dado:", valores.dtype)
```


# 🐼 Introdução ao Pandas: Manipulação de Dados em Python

O **Pandas** é uma biblioteca de código aberto da linguagem Python, amplamente utilizada nas áreas de **ciência de dados**, **engenharia de dados**, **inteligência artificial** e **análise estatística**. Seu principal objetivo é oferecer ferramentas poderosas e expressivas para a **manipulação, limpeza, análise e transformação de dados estruturados e semi-estruturados**.

O nome **"Pandas"** vem da expressão *Panel Data*, um termo da econometria que descreve conjuntos de dados multidimensionais que acompanham variáveis ao longo do tempo. A biblioteca foi desenvolvida por **Wes McKinney**, em 2008, durante seu trabalho com dados financeiros na empresa AQR Capital Management. McKinney percebeu que as ferramentas disponíveis na época para manipulação de dados em Python eram limitadas em relação ao que o setor financeiro exigia. A partir disso, o Pandas foi criado e evoluiu rapidamente, tornando-se uma das bibliotecas mais importantes do ecossistema de dados moderno.

Atualmente, o Pandas é utilizado por **pesquisadores, engenheiros, analistas de dados e cientistas** ao redor do mundo em tarefas que vão desde análises exploratórias simples até a preparação de grandes volumes de dados para **modelos de machine learning, inteligência artificial e sistemas de produção**.

---

## 📚 Por que o Pandas é importante?

O Pandas oferece estruturas de dados altamente otimizadas e intuitivas para trabalhar com dados tabulares:

- **`Series`**: são vetores unidimensionais rotulados, semelhantes a uma única coluna de uma planilha. Podem conter qualquer tipo de dado (inteiros, strings, floats, objetos, datas).
- **`DataFrame`**: são estruturas bidimensionais (tabelas) com rótulos de linhas e colunas, semelhantes a uma planilha do Excel, a uma tabela SQL ou a uma matriz de dados do R.

Essas estruturas possibilitam uma série de operações essenciais para análise de dados:

- 📥 **Leitura de dados**: a partir de arquivos CSV, Excel, JSON, banco de dados SQL, APIs e até arquivos remotos.
- 🧹 **Limpeza de dados**: remoção de dados duplicados, substituição de valores ausentes, tratamento de inconsistências.
- 🔁 **Transformações**: criação de novas colunas, normalização, conversão de tipos, aplicação de funções customizadas com `apply()`.
- 🔍 **Filtragem e seleção**: acesso rápido e eficiente a subconjuntos de dados usando operadores lógicos, condições e índices personalizados.
- 📊 **Estatísticas descritivas**: cálculo de média, desvio padrão, mediana, somatórios e outros resumos analíticos com facilidade.
- 📈 **Agrupamentos e agregações**: operações com `groupby()` para análises por categorias, turnos, setores industriais, etc.
- 🤝 **Integração com outras bibliotecas**: interoperável com `NumPy` (arrays), `Matplotlib` e `Seaborn` (visualização), `Scikit-learn` (machine learning), entre outras.

Além disso, o Pandas é **altamente compatível com ambientes industriais** que lidam com grandes volumes de dados operacionais, como sensores de fábrica, logs de produção, controle de energia, e análise de desempenho de máquinas.

Com sua interface simples e poderosa, o Pandas permite que tarefas complexas de engenharia e ciência de dados sejam executadas com poucas linhas de código, reduzindo erros, aumentando a produtividade e facilitando a documentação e reprodutibilidade dos experimentos.

## 1. 🧱 Estruturas de Dados do Pandas

O Pandas possui duas **estruturas de dados principais** que você usará constantemente:

### 🔹 `Series`: Vetor unidimensional com índice

Uma `Series` é como uma **lista ou vetor**, mas com **índices nomeados** (rótulos). Pode ser comparada a uma coluna de planilha Excel ou a uma coluna de um banco de dados.

```python
import pandas as pd

# Exemplo: consumo energético de máquinas
consumo = pd.Series([120.5, 130.2, 115.0], index=["Máquina A", "Máquina B", "Máquina C"])
print(consumo)
```

🔍 A Series permite acessar elementos usando índices nomeados ou posições:

```python
consumo["Máquina A"]      # Acesso por rótulo
consumo[0]                # Acesso por posição
```

🔹 DataFrame: Tabela bidimensional (linhas + colunas)
O DataFrame é a estrutura central do Pandas. Ele representa uma tabela com linhas e colunas rotuladas, como uma planilha ou um resultado de consulta SQL.

```python
dados = {
    "Máquina": ["A", "B", "C"],
    "Consumo (kWh)": [120.5, 130.2, 115.0],
    "Operando": [True, False, True]
}

df = pd.DataFrame(dados)
print(df)
```

Você pode acessar colunas como em um dicionário:

```python
df["Consumo (kWh)"]
```

Ou acessar linhas por índice com .loc[] (por rótulo) ou .iloc[] (por posição):

```python
df.loc[1]       # Linha com índice 1
df.iloc[2]      # Terceira linha
```

### 🔄 Comparação com outras estruturas do Python

| **Característica**         | **Lista (`list`)**      | **Dicionário (`dict`)**       | **NumPy Array (`ndarray`)**  | **Pandas Series**           | **Pandas DataFrame**            |
|----------------------------|--------------------------|--------------------------------|-------------------------------|------------------------------|----------------------------------|
| **Dimensionalidade**       | 1D                       | 1D (*chave-valor*)             | 1D ou 2D                      | 1D                           | 2D                               |
| **Suporte a rótulos**      | ❌ (somente índice)      | ✅ (chaves)                    | ❌                            | ✅ (índice nomeado)          | ✅ (índice + colunas)            |
| **Operações vetoriais**    | ❌                       | ❌                             | ✅                            | ✅                           | ✅                               |
| **Ideal para**             | listas simples           | dados dispersos                | cálculos numéricos rápidos   | séries de dados              | tabelas e análise tabular        |

## 2. 📥 Leitura e Escrita de Dados com Pandas

Um dos pontos fortes do Pandas é a capacidade de **ler e salvar dados de diversas fontes** com apenas uma linha de código. Isso permite integrar dados de planilhas, bancos, sensores, APIs e arquivos em texto de forma rápida e confiável.

### 🔹 Leitura de Arquivos com `read_csv()`

O método `read_csv()` é o mais comum e permite carregar arquivos no formato **CSV** (*Comma-Separated Values*), que é uma tabela de texto plano.

```python
import pandas as pd

df = pd.read_csv("dados_producao.csv")
```

No entanto, no Brasil, é comum que arquivos CSV usem:

Vírgula (,) como separador decimal nos números reais

Ponto e vírgula (;) como separador de colunas

Para esses casos, devemos configurar os parâmetros corretamente:

```python
# Exemplo para CSV brasileiro
df = pd.read_csv("dados_brasil.csv", sep=";", decimal=",", encoding="utf-8")
```

⚠️ Dica:
sep=";" → define o separador de colunas

decimal="," → define o separador decimal

encoding="utf-8" → define a codificação (padrão internacional)

## 3. 🧹 Limpeza e Transformação de Dados

Em projetos reais de ciência de dados e machine learning, **limpar e transformar os dados** é uma das etapas mais importantes e trabalhosas. Arquivos do mundo real frequentemente apresentam:

- Valores ausentes
- Formatos inconsistentes
- Dados duplicados
- Tipos de dados incorretos
- Colunas com nomes ambíguos

A biblioteca Pandas fornece várias ferramentas eficientes para resolver esses problemas de forma rápida e expressiva.

---

### 🔸 `dropna()` – Remover dados ausentes

Remove linhas ou colunas que contêm valores ausentes (NaN - *Not a Number*).

```python
import pandas as pd
import numpy as np

# Exemplo: produção por turno com dados faltando
df = pd.DataFrame({
    "Turno": ["Manhã", "Tarde", "Noite", "Madrugada"],
    "Produção (peças)": [150, 200, np.nan, 180]
})

# Remove a linha com valor ausente
df_limpo = df.dropna()
print(df_limpo)
```

🔧 Use axis=1 para remover colunas em vez de linhas.

🔸 fillna() – Preencher valores ausentes
Em vez de remover, podemos preencher os valores ausentes com um valor padrão ou com a média, mediana, etc.

```python
# Preenche com 0
df.fillna(0)

# Preenche com a média da coluna
media = df["Produção (peças)"].mean()
df.fillna(media)
```

🔸 replace() – Substituir valores específicos
Permite substituir valores errados, códigos, ou rótulos por novos significados.

```python
# Exemplo: status codificado
status = pd.Series(["OK", "X", "OK", "F", "X", "OK"])
status_corrigido = status.replace({"X": "Falha", "F": "Fora de padrão"})
print(status_corrigido)
```

🔸 Renomear colunas com .rename()
Nomes de colunas vindos de arquivos externos nem sempre são claros. Você pode renomeá-los com:

```python
df = df.rename(columns={
    "Produção (peças)": "Qtde Produzida",
    "Turno": "Turno de Trabalho"
})
```

📌 Ideal para padronizar nomes e facilitar o uso em funções e modelos.

🔸 Converter tipos com .astype()
Às vezes, colunas numéricas são lidas como texto ou datas como objetos. Podemos corrigir com:

```python
# Converte uma coluna para inteiro
df["Qtde Produzida"] = df["Qtde Produzida"].astype(int)

# Converte coluna de datas
df["Data"] = pd.to_datetime(df["Data"])
```

⚠️ Muito importante antes de alimentar modelos de machine learning!

🔸 Remover duplicatas com .drop_duplicates()
É comum que arquivos contenham linhas repetidas (por erro ou reprocessamento).

```python
df = pd.DataFrame({
    "Máquina": ["A", "B", "A", "C", "A"],
    "Produção": [100, 150, 100, 130, 100]
})

# Remove linhas idênticas
df_unico = df.drop_duplicates()
print(df_unico)
```

Você também pode especificar colunas a considerar:

```python
df.drop_duplicates(subset=["Máquina"])
```

### ✅ Resumo da Limpeza de Dados

| **Função**            | **Finalidade**                                              |
|------------------------|-------------------------------------------------------------|
| `dropna()`            | Remove linhas ou colunas com valores ausentes               |
| `fillna()`            | Preenche valores ausentes com valores definidos             |
| `replace()`           | Substitui valores específicos                                |
| `rename()`            | Renomeia colunas                                             |
| `astype()`            | Converte tipos de dados                                      |
| `drop_duplicates()`   | Remove linhas duplicadas                                     |

Essas funções permitem preparar os dados de forma adequada para análises estatísticas, visualizações ou entrada em modelos de Machine Learning. O Pandas garante que essas operações sejam feitas de forma segura, performática e legível.

## 4. ⚙️ Operações com Colunas

Um dos maiores poderes do Pandas está na facilidade de **criar, transformar e manipular colunas** de forma declarativa e eficiente. Nesta seção, vamos aprender:

- Como **criar novas colunas derivadas**
- Como aplicar **funções personalizadas** com `apply()` e `map()`

Essas técnicas são fundamentais em tarefas de **engenharia de atributos (feature engineering)** em projetos de Machine Learning e Análise de Dados.

---

### 🆕 Criação de Colunas Derivadas

Você pode criar novas colunas diretamente com base em operações sobre colunas existentes.

#### 🔹 Exemplo: Eficiência Energética

Vamos supor que temos um DataFrame com consumo de energia e produção de peças.

```python
import pandas as pd

df = pd.DataFrame({
    "Equipamento": ["Torno A", "Prensa B", "Esteira C"],
    "Consumo (kWh)": [120.5, 150.3, 95.0],
    "Produção (peças)": [300, 400, 250]
})

# Criar nova coluna: consumo por peça
df["kWh por peça"] = df["Consumo (kWh)"] / df["Produção (peças)"]

print(df)
```

🧮 Operações Aritméticas com múltiplas colunas

```python
# Adicionando um custo estimado por peça (R$ 0,75 por kWh)
df["Custo por peça (R$)"] = df["kWh por peça"] * 0.75
```

🔄 Criação com expressões condicionais

```python
# Classificar desempenho energético
df["Desempenho"] = df["kWh por peça"].apply(
    lambda x: "Alta Eficiência" if x < 0.4 else "Regular" if x < 0.6 else "Baixa Eficiência"
)
```

🧠 Aplicação de Funções com apply() e map()

🔧 .apply() – Aplica uma função a cada linha ou coluna

🔹 Aplicar função a uma coluna inteira:

```python
# Converter nomes para maiúsculas
df["Equipamento"] = df["Equipamento"].apply(str.upper)
```

🔹 Aplicar função a linhas inteiras (axis=1)

```python
# Criar etiqueta completa da linha
df["Etiqueta"] = df.apply(lambda row: f"{row['Equipamento']} - {row['Desempenho']}", axis=1)
```

🔁 .map() – Para transformar valores de uma coluna individual

É mais simples e rápido que apply() para mapeamentos diretos.

```python
status = pd.Series(["OP", "OP", "MA", "PA", "OP"])
# Traduzir siglas para nomes completos
status_map = {"OP": "Operacional", "MA": "Manutenção", "PA": "Parado"}
status = status.map(status_map)
print(status)
```

🧠 Comparação entre apply() e map()

| **Método** | **Aplica a**     | **Uso ideal**                                    |
|------------|------------------|---------------------------------------------------|
| `map()`    | Coluna           | Substituições simples ou rótulos                 |
| `apply()`  | Linha ou coluna  | Lógicas complexas ou múltiplas colunas           |

📌 Outros exemplos úteis com apply()
Calcular faixas de risco com base em múltiplas colunas:

```python
def risco(temp, status):
    if temp > 90 and status == "Manutenção":
        return "⚠️ Risco Alto"
    elif temp > 80:
        return "Risco Médio"
    else:
        return "Risco Baixo"

df["Risco Operacional"] = df.apply(lambda row: risco(row["kWh por peça"], row["Desempenho"]), axis=1)
```

🧩 Uso em IA: engenharia de atributos

Na construção de modelos preditivos, você pode:

Criar colunas numéricas normalizadas

Converter rótulos em números (map + replace)

Gerar colunas booleanas com base em regras (apply)

Criar colunas de data/hora derivadas (mês, hora, etc.)

✅ Resumo
| **Ação**                         | **Método** | **Exemplo**                                            |
|----------------------------------|------------|--------------------------------------------------------|
| Criar coluna com operação        | `df[...] =`| `df["A"] / df["B"]`                                      |
| Aplicar função a coluna          | `apply()`  | `df["col"].apply(minha_funcao)`                        |
| Aplicar função a linha inteira   | `apply()`  | `df.apply(lambda row: ..., axis=1)`                    |
| Substituir valores diretamente   | `map()`    | `df["status"].map(dicionario_de_substituição)`         |

Essas técnicas tornam o Pandas uma ferramenta poderosa para pré-processamento de dados, etapa essencial em qualquer pipeline de Machine Learning industrial ou empresarial.

## 🔍 Seleção, Filtros e Reordenação de Colunas

Manipular colunas é essencial para limpar, organizar e preparar os dados antes de qualquer análise ou aplicação de modelos. Nesta seção veremos como:

- Selecionar uma ou mais colunas
- Filtrar linhas com base em condições
- Reordenar as colunas do DataFrame

---

### 🔸 Seleção de colunas

#### ✅ Selecionar uma única coluna

```python
df["Equipamento"]         # retorna uma Series
```

✅ Selecionar múltiplas colunas

```python
df[["Equipamento", "Produção (peças)"]]  # retorna um novo DataFrame
```

✅ Acessar colunas como atributos (quando não há espaços ou acentos)

```python
df.Equipamento
```

🔸 Filtro de linhas com base em condições
Exemplo: selecionar apenas os equipamentos que produziram mais de 300 peças

```python
df[df["Produção (peças)"] > 300]
```

Filtro com múltiplas condições

```python
df[(df["Produção (peças)"] > 300) & (df["Status"] == "Operacional")]
```

⚠️ Use & para "E" e | para "OU", sempre com parênteses.

🔸 Filtro com isin() – útil para categorias

```python
# Selecionar apenas linhas com determinados equipamentos
df[df["Equipamento"].isin(["Torno A", "Prensa B"])]

# Filtrar nomes que contenham "FORNO" (insensível a maiúsculas)
df[df["Equipamento"].str.contains("FORNO", case=False)]
```

🔸 Filtro com valores ausentes

```python
# Selecionar apenas as linhas onde a temperatura está ausente
df[df["Temperatura (°C)"].isna()]
```

🔄 Reordenação de colunas

Às vezes queremos reorganizar a ordem das colunas para facilitar a leitura ou exportação:

```python
# Definir nova ordem
nova_ordem = ["Timestamp", "Equipamento", "Status", "Produção (peças)", "Temperatura (°C)"]
df = df[nova_ordem]
```

✅ Dicas de boas práticas

Use seleção de colunas para reduzir a memória utilizada

Combine filtros para refinar análises

Reordene colunas antes de exportar dados limpos

Filtre colunas específicas antes de aplicar .describe() ou .groupby()

Essas técnicas são fundamentais para limpeza, segmentação e exportação de dados.

### 🗑️ Explicação do método `drop()`

O método `drop()` é utilizado para **remover colunas ou linhas** de um `DataFrame`.

- Para remover colunas, use `axis=1`
- Para remover linhas, use `axis=0` (ou informe diretamente o índice com `index=`)

#### 🔹 Remover uma coluna
```python
df = df.drop("Status", axis=1)
```

#### 🔹 Remover múltiplas colunas
```python
df = df.drop(["Temperatura (°C)", "Consumo (kWh)"], axis=1)
```

#### 🔹 Remover uma linha pelo índice
```python
df = df.drop(index=3)
```

#### 🔹 Remover várias linhas pelo índice
```python
df = df.drop(index=[0, 2, 4])
```

O `drop()` não altera o `DataFrame` original a menos que `inplace=True` seja especificado.

```python
df.drop("Status", axis=1, inplace=True)
```

## 🔍 5. Filtragem e Seleção de Dados

Na verdade quando tratamos de operações com colunas já falamos um pouco sobre filtragem, pois utilizamos os nomes ou números das colunas para trabalhar com as seleções de dados, mas vamos revisar isso e nos focarmos na questão de filtrar e selecionar dados com precisão. Esta operação é essencial.

O Pandas fornece duas ferramentas poderosas para isso:

- `loc[]`: seleção por **rótulo** (label)
- `iloc[]`: seleção por **posição** (index numérico)

### 🔹 `loc[]` – Acesso por rótulo

Permite selecionar **linhas e colunas nomeadas**:

```python
# Selecionar linha com índice 3
linha3 = df.loc[3]

# Selecionar múltiplas linhas pelo índice
df.loc[[1, 4, 6]]

# Selecionar intervalo de linhas e colunas nomeadas
df.loc[0:5, ["Equipamento", "Produção (peças)"]]
```

Também permite filtragens com condições:

```python
# Selecionar linhas onde a produção foi superior a 300
df.loc[df["Produção (peças)"] > 300]
```

---

### 🔸 `iloc[]` – Acesso por posição

Usa **índices inteiros**, como listas ou arrays NumPy.

```python
# Linha de índice 2
df.iloc[2]

# Linhas 0 a 4 e colunas 0 a 2
df.iloc[0:5, 0:3]

# Última linha
df.iloc[-1]
```

---

### 🔁 Filtros condicionais e lógicos

Filtragens com expressões booleanas permitem consultas refinadas:

```python
# Equipamentos com produção > 300 e desempenho regular
df[(df["Produção (peças)"] > 300) & (df["Desempenho"] == "Regular")]

# Equipamentos que não são Prensa B
df[df["Equipamento"] != "Prensa B"]
```

Também podemos utilizar `isin()`:

```python
# Selecionar linhas com certos equipamentos
df[df["Equipamento"].isin(["Torno A", "Forno C"])]
```

Ou `str.contains()` para buscas em colunas de texto:

```python
# Linhas onde o nome do equipamento contém "FORNO"
df[df["Equipamento"].str.contains("FORNO", case=False)]
```

Essas técnicas são fundamentais para extrair subconjuntos relevantes de dados para análise estatística, visualização ou modelagem preditiva.

## 📊 6. Agrupamentos e Estatísticas

A análise de dados industriais geralmente requer **agregar e comparar métricas** por categoria, equipamento, turno, tipo de processo, dia da semana, entre outros. Isso é fundamental para identificar gargalos, prever falhas e otimizar recursos.

O Pandas oferece ferramentas como `groupby()`, `agg()` e `pivot_table()` que permitem **agregações poderosas, flexíveis e altamente customizáveis**, sendo essenciais para sumarização e visualização de dados complexos.

### 🔹 `groupby()` – Agrupar linhas por valores em uma ou mais colunas

```python
# Média da produção por equipamento
df.groupby("Equipamento")["Produção (peças)"].mean()

# Total de consumo por turno
df.groupby("Turno")["Consumo (kWh)"].sum()

# Quantidade de registros por tipo de equipamento
df.groupby("Equipamento").size()
```

Você pode agrupar por múltiplas colunas:

```python
# Média de produção por turno e equipamento
df.groupby(["Turno", "Equipamento"])["Produção (peças)"].mean()
```

### 🔹 `agg()` – Agregações múltiplas por grupo

```python
# Média e desvio padrão por equipamento
df.groupby("Equipamento")["Produção (peças)"].agg(["mean", "std"])
```

#### 📌 Exemplos adicionais com `agg()`

```python
# Agregações múltiplas com renomeação de colunas
df.groupby("Turno")["Consumo (kWh)"].agg(
    Média_kWh="mean",
    Máximo_kWh="max",
    Mínimo_kWh="min"
)

# Agrupamento com mais de uma coluna numérica
df.groupby("Equipamento")[["Produção (peças)", "Consumo (kWh)"]].agg(["mean", "sum"])
```

Você também pode usar funções customizadas:

```python
def faixa(x):
    return x.max() - x.min()

df.groupby("Equipamento")["Produção (peças)"].agg(["mean", faixa])
```

### 🔹 Agregação em múltiplas colunas

```python
# Média de produção e consumo por equipamento
df.groupby("Equipamento")[["Produção (peças)", "Consumo (kWh)"]].mean()
```

### 🔸 Pivot Tables – Tabelas dinâmicas

As pivot tables em Pandas são semelhantes às do Excel e permitem **agregar e reorganizar dados** em forma de matriz. São úteis para gerar relatórios dinâmicos, sumarizar grande volume de dados e visualizar comparações cruzadas.

```python
# Soma da produção por turno e equipamento
pd.pivot_table(
    df,
    values="Produção (peças)",
    index="Turno",
    columns="Equipamento",
    aggfunc="sum"
)
```

#### 📌 Outras agregações com `pivot_table()`:

```python
# Média de consumo por turno e tipo de equipamento
pd.pivot_table(
    df,
    values="Consumo (kWh)",
    index="Turno",
    columns="Tipo",
    aggfunc="mean"
)
```

Você pode usar `margins=True` para obter totais gerais:

```python
pd.pivot_table(
    df,
    values="Produção (peças)",
    index="Turno",
    columns="Equipamento",
    aggfunc="sum",
    margins=True
)
```

### ✅ Resumo das funções úteis

| Ação                                              | Função                       |
|---------------------------------------------------|------------------------------|
| Agrupar e calcular média                          | `groupby()` + `mean()`       |
| Múltiplas estatísticas por grupo                  | `agg()`                      |
| Agregações customizadas com funções definidas     | `agg()` com função lambda    |
| Agregações cruzadas em forma tabular              | `pivot_table()`              |
| Obter totais gerais em pivôs                      | `pivot_table(..., margins=True)` |
| Contagem de registros por grupo                   | `groupby().size()`           |
| Desvio padrão de um grupo                         | `groupby()` + `std()`        |
